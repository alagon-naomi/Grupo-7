 **  Sistema Bancario 
 * (el “Banco Nacional del Ahorro”) está completo y lo probé: funciona.
Qué hace el sistema:

1.	Registra clientes, validando que el DNI tenga 8 dígitos y no esté repetido.

2.	Abre dos tipos de cuenta:

o	Cuenta de Ahorros: permite como máximo 3 retiros al mes y gana 1 % de interés.

o	Cuenta Corriente: permite sobregiro de hasta S/ 500 y cobra S/ 10 de mantenimiento.
3.	Hace depósitos, retiros y transferencias.
4.	Muestra el estado de cuenta con todos los movimientos.
5.	Hace el cierre de mes y un reporte general del banco.
